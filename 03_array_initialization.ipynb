# NumPy — Array Initialization Methods


```python
import numpy as np
```

```python
# zeros
zero_arr = np.zeros((2,3))
print(zero_arr)
```

```python
# ones
ones_arr = np.ones((2,3))
print(ones_arr)
```

```python
# full
full_arr = np.full((3,2), 7)
print(full_arr)
```

```python
# identity matrix
iden_arr = np.eye(3)
print(iden_arr)
```

```python
# empty array
arr = np.empty(2)
print(arr)
```

```python
# evenly spaced integer values
print(np.arange(1,3,1))
```

```python
# equally spaced values in a range
print(np.linspace(1,10,5))
print(np.linspace(1,10,4))
```

```python
# random float values
random_arr = np.random.rand(3,2)
print(random_arr)
```

```python
# random integer values
random_arr1 = np.random.randint(1,100,(3,3))
print(random_arr1)
```
