# NumPy — Mathematical Operations on Arrays


```python
import numpy as np
```

```python
a = np.array([10,20,-30,40])
print(a)
```

```python
print(a+20)
print(a-30)
print(a*2)
print(a/10)
```

```python
b = np.array([1,4,9])
print(np.square(b))
print(np.sqrt(b))
```
