# NumPy — Array Reshaping and Flattening


```python
import numpy as np
```

```python
# Reshaping
arr1 = np.array([1,2,3,4,5])
print(arr1)
reshaped = arr1.reshape((5,1))
print(reshaped)
```

```python
arr1 = np.array([[1,2,3,4,5,6]])
reshaped2 = arr1.reshape(2,3)
print(reshaped2)
```

```python
# Flattening
arr2 = np.array([[1,2,3,4,5,6]])
print(arr2)
print(arr2.flatten())
```
