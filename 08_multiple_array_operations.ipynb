# NumPy — Mathematical Operations on Multiple Arrays


```python
import numpy as np
```

```python
a = np.array([1,2,3])
b = np.array([4,5,6])
print(a)
print(b)
```

```python
print(np.add(a,b))
print(np.subtract(a,b))
print(np.divide(a,b))
print(np.multiply(a,b))
```

```python
# Dot product
print(np.dot(a,b))
```
