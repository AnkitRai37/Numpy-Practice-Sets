```python
import numpy as np
```

```python
arr1 = np.array([1,3,2,7,2,6,9,70,10,20,49,6,324,35])
print(arr1)
```

```python
# 1.1 Boolean condition - doesn't directly give you the values. It creates a Boolean mask.
print(arr1 > 9) 
```

```python
print(arr1 == 70)
print(arr1 != 49)
print(arr1 > 40)
print(arr1 < 60)
print(arr1 <= 30)
print(arr1 >= 35)
```

```python
# 1.2 True / False Mask - A mask is simply an array of True and False.
print(arr1 > 40)
```

```python
print(arr1[arr1 > 60])
```

```python
# 1.3 Filtering Values- This is essentially filtering rows based on a condition, which you'll use heavily later in Pandas.
sales = np.array([1200,356,2000,480,4500,3421,857])
print(sales)
```

```python
print(sales[sales > 2000])
```

```python
print(sales[sales < 1000])
```

```python
print(sales[sales == 480])
```

```python
# 1.4 Multiple Conditions - They are important when combining conditions.
arr3 = np.array([100,0,2341,3412,647,2421,9080])
print(arr3)
```

```python
print(arr3[(arr3 >= 2000) & (arr3 <= 5000)])
```

```python
#1.5, AND - & means both conditions must be True.
arr4 = np.array([1000,2346,8643,5735,1231,4326])
print(arr4)
```

```python
print(arr3[(arr3 > 2000) & (arr3 < 6000)])
```

```python
# 1.6, — OR- | means at least one condition must be True.
arr5 = np.array([91208,4325,352,8528,5646,1231,3242,5467])
print(arr5)
```

```python
print(arr5[(arr5 > 3000) | (arr5 < 1000 )])
```

```python
# 1.7, — NOT- ~ reverses a Boolean condition.
arr6 = np.array([123,3242,4535,3473,1285,9267,3528])
print(arr6)
```

```python
print(arr6[arr6 > 3000])
```

```python
print(arr6[~(arr6 > 3000)])
```

```python
# 2. Fancy Indexing - Fancy indexing means selecting elements using an array/list of indices.
```

```python
# 2.1 Selecting Specific Positions
arr = np.array([10,20,30,40,50])
print(arr)
```

```python
print(arr[[0,1,4]])  # selects multiple positions.
```

```python
print(arr[2])  # selects one position.
```

```python
#2.2 Selecting Multiple Rows - For a 2D array: - So fancy indexing isn't restricted to sequential positions.
data = np.array([
    [101, 500],
    [102, 700],
    [103, 900],
    [104, 1200]
])
print(data[[0,3]])
```

```python
print(data[[3,1,0]]) # You can also select rows in any order:
```

```python
# 2.3 Selecting Multiple Columns
data = np.array([
    [101, 500, 10],
    [102, 700, 20],
    [103, 900, 30]
])
print(data)
```

```python
print(data[ : ,[0,2]])
```

```python
# 2.4 Combining Row & Column Selection
data = np.array([
    [101,5000,10],
    [102,700,95],
    [103,458,548],
    [104,215,672]])
print(data)
```

```python
data[np.ix_([0, 2], [1, 2])] # - It creates the proper row/column combination for selecting a rectangular subset.
# np.ix_(row_indices, column_indices)
```

```python
# 3. Boolean + Fancy Indexing
data = np.array([[10,748,258],[102,420,103],[45,302,857],[100,20,85],[40,58,41],[18,52,790]])
print(data)
```

```python
#  syntax - data[condition, column]
# Find rows where sales > 1000
data[:,1] > 150 #boolean / mask and 1 is column
```

```python
data[data[:,1] > 150] # - 1 is colmnn 1
```

```python
data[data[:,1] > 150, 1] # first 1 is column 1 giving input and sec 1 is returning column 1.
```

```python
# Multiple conditions on rows
data[(data[:,1] >=200) & (data[:,2] <= 200)]
```

```python
#, Practical Data Analyst Example
sales = np.array([
    [101, 1200, 5],
    [102, 800,  3],
    [103, 2500, 8],
    [104, 1500, 6],
    [105, 500,  2]
])
print(sales)
```

```python
# Products with sales > 1000
print(sales[sales[:,1] > 1000]) # 1 is sales column
```

```python
# - Product IDs with sales > 1000
print(sales[sales[:,1]>1000,0])
```

```python
# - Products with sales > 1000 AND quantity > 5
print(sales[(sales[:,1] > 1000) & (sales[:,2] > 5)])
```

```python
# Product IDs satisfying both conditions
print(sales[(sales[:,1] > 1000) & (sales[:,2] > 5), 0]) # 0 is 1st column
```
