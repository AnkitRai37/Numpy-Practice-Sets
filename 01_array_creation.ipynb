# NumPy — Array Creation


```python
import numpy as np
```

```python
# 1D array
arr1 = np.array([1,2,3])
print(arr1)
```

```python
# 2D array
list1 = [1,2,3]
list2 = [4,5,6]
arr2 = np.array([list1, list2])
print(arr2)
```

```python
# 3D array
list1 = [1,2,3]
list2 = [4,5,6]
list3 = [7,8,9]
arr3 = np.array([[list1, list2, list3]])
print(arr3)
```

```python
# Number of dimensions
print(arr3.ndim)
```
