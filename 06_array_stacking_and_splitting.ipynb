# NumPy — Array Stacking and Splitting


```python
import numpy as np
```

```python
# Stacking
a = np.array([1,2,3])
b = np.array([4,5,6])
print(np.vstack((a,b)))
print(np.hstack((a,b)))
```

```python
# Splitting
c = np.array([[1,2,3],[4,5,6]])
print(c)
```

```python
hsplit = np.hsplit(c,3)
for s in hsplit:
    print(s)
```

```python
vsplit = np.vsplit(c,2)
for s in vsplit:
    print(s)
```
