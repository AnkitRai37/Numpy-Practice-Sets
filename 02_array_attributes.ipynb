# NumPy — Array Attributes


```python
import numpy as np
```

```python
arr1 = np.array([[1,2,3],[4,5,6]])
print(arr1)
```

```python
print(np.shape(arr1))
print(np.size(arr1))
print(arr1.dtype)
print(np.ndim(arr1))
```

```python
# Attribute form
print(arr1.shape)
print(arr1.size)
print(arr1.dtype)
print(arr1.ndim)
```
